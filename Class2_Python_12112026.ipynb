```python
#STRING
#Operators:
#Concatination
first_name = "Virat"
last_name = "rahul"
print( first_name + "-" + last_name)
print("-" * 30)
print("My firstname is " + first_name + " and my last name is " + last_name)
print("-" * 30)

#Repetation

print(first_name * 10)

#Membership

statement = "Yesterday arun went to a icecream shop"
print("arun" in statement)
print("Gopi" in statement)
print("ice" in statement)

#Not a Membership
statement = "Yesterday arun went to a icecream shop"
print("arun" not in statement)
print("Gopi" not in statement)
print("ice" not in statement)

#Comparision
name_1 = "arun"
name_2 = "arun"
print(name_1 == name_2)
#---------------------------------
# Slicing , methods, fucntions
# --------------------------------
```

```python
# SLICING
# sequence[start:stop:step]
# It will start with 0th Index, and capturing between datas form index.

text = "PYTHON"

print(text[0:3]) #1,2 INDEX WILL BE CAPTURED
print(text[2:5]) #2,3,4 INDEX WILL BE CAPTURED
print(text[:4]) #0,1,2,3 INDEX WILL BE CAPTURED
print(text[2:4]) #2,3 INDEX WILL BE CAPTURED
print(text[::-1]) # REVERSE THE STRING
```
